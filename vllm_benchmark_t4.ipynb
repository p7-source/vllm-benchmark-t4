```python
import time
from vllm import LLM, SamplingParams

llm = LLM(model="Qwen/Qwen2.5-1.5B-Instruct", dtype="half", gpu_memory_utilization=0.8)
params = SamplingParams(max_tokens=128, temperature=0.8)
prompts = [f"Explain GPUs in two sentences. #{i}" for i in range(128)]

for batch in [1, 8, 32, 128]:
    start = time.time()
    outputs = llm.generate(prompts[:batch], params)
    secs = time.time() - start
    tokens = sum(len(o.outputs[0].token_ids) for o in outputs)
    print(f"batch={batch:4d}  time={secs:6.2f}s  throughput={tokens/secs:8.1f} tok/s")
```

```python
# Phase 2: Profiling - find where GPU time goes
# Setup:  pip install torch     (GPU)
import torch
from torch.profiler import profile, ProfilerActivity, record_function

model = torch.nn.Sequential(
    torch.nn.Linear(4096, 4096), torch.nn.ReLU(), torch.nn.Linear(4096, 4096)
).cuda()
x = torch.randn(64, 4096, device="cuda")

for _ in range(5):          # warm-up so we don't profile one-time setup
    model(x)
torch.cuda.synchronize()

with profile(activities=[ProfilerActivity.CPU, ProfilerActivity.CUDA], record_shapes=True) as prof:
    with record_function("forward_pass"):
        for _ in range(10):
            model(x)
    torch.cuda.synchronize()

for key in ("device_time_total", "cuda_time_total"):   # name differs by torch version
    try:
        print(prof.key_averages().table(sort_by=key, row_limit=10)); break
    except Exception:
        pass
prof.export_chrome_trace("trace.json")
print("Open trace.json in https://ui.perfetto.dev to see the timeline.")
# Try next: model.half() or torch.compile(model) and compare the top kernels.
```

```python
import torch
from torch.profiler import profile, ProfilerActivity, record_function

def run(dtype, name):
    model = torch.nn.Sequential(
        torch.nn.Linear(4096, 4096), torch.nn.ReLU(), torch.nn.Linear(4096, 4096)
    ).cuda().to(dtype)
    x = torch.randn(64, 4096, device="cuda", dtype=dtype)
    for _ in range(5):
        model(x)
    torch.cuda.synchronize()
    with profile(activities=[ProfilerActivity.CPU, ProfilerActivity.CUDA]) as prof:
        with record_function("forward_pass"):
            for _ in range(10):
                model(x)
        torch.cuda.synchronize()
    print(f"\n===== {name} =====")
    print(prof.key_averages().table(sort_by="cuda_time_total", row_limit=5))
    prof.export_chrome_trace(f"trace_{name}.json")

run(torch.float32, "fp32")
run(torch.float16, "fp16")
```
